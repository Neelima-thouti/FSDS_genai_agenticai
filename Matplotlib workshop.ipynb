```python
import pandas as pd

data={
    'Month':['Jan','Feb','Mar','Apr','May','Jun'],
    'Sales':[10000,12000,15000,13000,17000,16000],
    'Profit':[2000,3000,4000,2500,3500,3000]}

df=pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.plot(df['Month'],df['Sales'],color="blue",marker='o',ls='-',label='Sales')
plt.title('Sales Trends over Months')
plt.xlabel('Month')
plt.ylabel('Sales($)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,5))
width=0.3
plt.bar(df['Month'],df['Sales'],width=width,color="skyblue",ls='-',label='Sales')
plt.bar(df['Month'],df['Profit'],width=width,color="green",ls='-',label='Profit',bottom=df["Sales"])
plt.title('Sales and Profit Trends over Months')
plt.xlabel('Month')
plt.ylabel('Amount($)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from enum import auto
plt.figure(figsize=(10,4))
plt.pie(df['Profit'], labels=df['Month'],autopct='%1.2f%%',startangle=140,colors=plt.cm.Paired.colors)
plt.title('Profit by month')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Sales'],df['Profit'],color='red',s=100,edgecolors='black'),
plt.title('Sales Vs Profit Scatter Plot'),
plt.xlabel('Sale'),
plt.ylabel('Profit'),
plt.tight_layout(),
plt.grid(True),
plt.show(),
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Sales'],bins=5,color='purple',edgecolor='black'),
plt.title('Sales Distribution'),
plt.xlabel('Sale'),
plt.ylabel('Frequency'),
plt.tight_layout(),
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.boxplot(df['Profit'],vert=False,patch_artist=True,boxprops=dict(facecolor='lightgreen')),
plt.title('Sales Box Plot'),
plt.ylabel('Sales'),
plt.tight_layout(),
plt.show()
```

```python
import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt
data={
'Month':['Jan','Feb','Mar','Apr','May','Jun'],
'Sale':[10000,12000,15000,13000,17000,16000],
'Profit':[2000,3000,4000,2500,3500,3000]}
df=pd.DataFrame(data)
def generate_plot(plot_type):
    fig=plt.figure(figsize=(8,5))
    if plot_type=='Line Plot':
      plt.plot(df['Month'],df['Sale'],color='blue',marker='o',ls='-',label='Sales')
      plt.title('Sale over Month')
      plt.xlabel('Month')
      plt.ylabel('Sale')
      plt.legend()
      plt.grid(False)     
    elif plot_type=='stacked bar chart':
      fig.set_size_inches(10,6)
      width=0.3
      plt.bar(df['Month'],df['Sale'],width=width,color='skyblue',label='Sale')
      plt.bar(df['Month'],df['Profit'],width=width,color='green',label='Profit',bottom=df["Sales"])
      plt.title('Sale and Profit Trends by Monthly')
      plt.xlabel('Month')
      plt.ylabel('Amount($)')
      plt.grid(True)
      plt.legend()
    elif plot_type=='Pie chart':
       fig.set_size_inches(7,7)
       plt.figure(figsize=(10,4))
       plt.pie(df['Profit'], labels=df['Month'],autopct='%1.2f%%',startangle=140,colors=plt.cm.Paired.colors)
       plt.title('Profit by month')
    elif plot_type=='Scatter plot':
       plt.scatter(df['Sale'],df['Profit'],color='red',s=100,edgecolors='black')
       plt.title('Sales Vs Profit Scatter Plot')
       plt.xlabel('Sale')
       plt.ylabel('Profit')
       plt.grid(True)
    elif plot_type=='Histogram':
      plt.hist(df['Sale'],bins=5,color='purple',edgecolor='black')
      plt.title('Sales Distribution')
      plt.xlabel('Sale($)')
      plt.ylabel('Frequency')
    elif plot_type=='Box plot':
      plt.figure(figsize=(8,5))
      plt.boxplot(df['Profit'],vert=False,patch_artist=True,boxprops=dict(facecolor='lightgreen')),
      plt.title('Sales Box Plot'),
      plt.ylabel('Sales'),
      plt.tight_layout(),
      
    plt.tight_layout()
    return fig
  
  # gradio ui
demo= gr.Interface(
    fn=generate_plot,
    inputs=gr.Radio(
        ['Line Plot','stacked bar chart','Pie Chart','Scatter Plot','Histogram','Box plot'],
        label='choose plot type'),
    
outputs=gr.Plot(label='Visualization'),
title='Sales & Profit Visual Insight',
description='choose the type of visualize data'
)
demo.launch()
  
      
```
